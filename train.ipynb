```python
import pandas as pd
import numpy as np
```

```python
# 학습 데이터 로드 
df = pd.read_csv("train.csv", encoding='utf-8-sig')
```

```python

# 결측치 처리 1: 50% 이상 결측치 피처 제거
threshold = 0.5
total_samples = len(df)

# 결측 비율 계산
missing_ratio = df.isnull().mean()

# 50% 이상인 피처 식별
cols_to_drop = missing_ratio[missing_ratio >= threshold].index.tolist()

print(f"제거할 피처 ({len(cols_to_drop)}개): {cols_to_drop}")
print(f"제거 전 shape: {df.shape}")

# 피처 제거
df_clean = df.drop(columns=cols_to_drop)

print(f"제거 후 shape: {df_clean.shape}")
```

```python
# 결측치 처리 2: 특정 피처 결측치 처리1
print("\n=== 2. 특정 피처 결측치 처리 ===")

# 1. nationality: 최빈값(내국인)으로 대체
mode_nationality = df_clean['nationality'].mode()[0]  # 첫 번째 최빈값
df_clean['nationality'].fillna(mode_nationality, inplace=True)
print(f"nationality 최빈값 '{mode_nationality}'으로 1개 대체")

# 2. completed_semester: 중앙값으로 대체 (이상치 고려)
median_semester = df_clean['completed_semester'].median()
df_clean['completed_semester'].fillna(median_semester, inplace=True)
print(f"completed_semester 중앙값 {median_semester}으로 28개 대체")

# 확인
print("\n=== 처리 후 결측 확인 ===")
print(df_clean[['nationality', 'completed_semester']].isnull().sum())
print(f"\n최종 shape: {df_clean.shape}")
```
